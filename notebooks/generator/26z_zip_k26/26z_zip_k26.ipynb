# 26z — paket output kernel 26 menjadi satu zip (CPU, tanpa GPU)


```python
# CPU only: pack kernel 26 candidate images, masks, records and G_B EMA weights into one zip for a single download.
from pathlib import Path
import zipfile, hashlib, json
INPUT=Path('/kaggle/input'); OUT=Path('/kaggle/working')
roots=[p.parent for p in INPUT.rglob('summary.json') if (p.parent/'candidates').exists()]
assert len(roots)==1, roots
R=roots[0]; s=json.loads((R/'summary.json').read_text()); assert s['test_used'] is False
files=[p for p in R.rglob('*') if p.is_file() and (p.suffix in ('.png','.json','.log') or p.name.endswith('_G_B_ema.pth'))]
with zipfile.ZipFile(OUT/'k26_result.zip','w',zipfile.ZIP_STORED) as z:
    for p in sorted(files): z.write(p, p.relative_to(R.parent))
h=hashlib.sha256((OUT/'k26_result.zip').read_bytes()).hexdigest()
(OUT/'k26_result.sha256').write_text(h+'\n')
print(len(files),'files',round((OUT/'k26_result.zip').stat().st_size/2**20),'MB',h)
```
